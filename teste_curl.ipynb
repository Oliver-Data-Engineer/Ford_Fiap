```python
import random
from curl_cffi import requests

def obter_lista_proxies():
    url_lista = "https://raw.githubusercontent.com/wiki/gfpcom/free-proxy-list/lists/http.txt"
    try:
        resp = requests.get(url_lista, timeout=10)
        if resp.status_code == 200:
            return [linha.strip() for linha in resp.text.splitlines() if linha.strip()]
    except Exception as e:
        print(f"Erro ao baixar lista de proxies: {e}")
    return []

proxys = obter_lista_proxies()

if not proxys:
    print("Nenhum proxy encontrado.")
    exit()

# Seleciona o proxy sem duplicar o protocolo
raw_proxy = random.choice(proxys)
proxy_escolido = raw_proxy if raw_proxy.startswith(("http://", "https://", "socks5://")) else f"http://{raw_proxy}"

url = "https://www.carrosnaweb.com.br/catalogo.asp?fabricante=Ford&modelo=Ranger"

try:
    print(f"Testando o proxy: {proxy_escolido}")
    resp = requests.get(url, impersonate="chrome", proxy=proxy_escolido, timeout=15.0)
    print(resp.status_code, resp.url)
except Exception as e:
    print(f"Falha na requisição com o proxy {proxy_escolido}: {e}")
```
