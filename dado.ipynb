```python
from sqlalchemy import create_engine, MetaData

# 1. Configura a conexão com o banco de dados local
engine = create_engine("sqlite:///ford_fiap.db", echo=False)

# 2. Carrega a estrutura das tabelas existentes (Reflection)
metadata = MetaData()
metadata.reflect(bind=engine)

# 3. Lista o nome de todas as tabelas presentes no banco de dados
tabelas = metadata.tables.keys()

print("Tabelas encontradas no banco de dados:")
for nome_tabela in tabelas:
    print(f"- {nome_tabela}")
```
